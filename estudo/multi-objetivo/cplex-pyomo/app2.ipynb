```python
import pandas as pd

retornos = pd.DataFrame({
    'PETR4': [ 1.0, -2.0,  3.0,  0.0,  2.0],
    'VALE3': [ 2.0, -1.0,  1.0,  1.0,  2.0],
})
cov = retornos.cov()         
arr = cov.to_numpy()
n = len(cov)

sigma = {(a, b): arr[a, b]    
         for a in range(n)
         for b in range(n)} 
```

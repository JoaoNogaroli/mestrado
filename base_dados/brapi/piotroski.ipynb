```python
import os
from brapi import Brapi
from dotenv import load_dotenv
import pandas as pd
import requests
import json

from pathlib import Path
```

```python
# DRE
#   - lucro liq = netIncome
#   - depreciacao = otherOperatingExpenses + costOfRevenue
#   - Margem bruta = grossProfit / totalRevenue
#   - Giro ativos = totalRevenue / ativo total(bp)

# Bp
#   - Ativo total = Cash + shortTermInvestments
#   - Clientes = netReceivables
#   - Estoques = inventory
#   - Fornecedores = providers
#   - Dívida = longTermLoansAndFinancing
#   - Liquidez Corrente  = totalCurrentAssets / currentLiabilities
#   - Giro ativos = totalrevenue(dre) / totalAssets


# Estatística
#   - Diluição = sharesOutstanding
```

```python


lista_df = []
for filename in os.listdir('dre'):
    
    # 2. Reconstruct the full absolute or relative path to the file
    try:
        full_path_balanco = os.path.join('balanco', filename)
        full_path_dre = os.path.join('dre', filename)
        full_path_estatistica = os.path.join('estatistica', filename)
        full_path_caixa = os.path.join('caixa', filename)
        full_path_financeiros = os.path.join('financeiros', filename)


        df_balanco = pd.read_csv(full_path_balanco)
        df_dre = pd.read_csv(full_path_dre)
        df_estatistica = pd.read_csv(full_path_estatistica)
        df_caixa = pd.read_csv(full_path_caixa)
        df_financeiro = pd.read_csv(full_path_financeiros)


        df_dre = df_dre[['endDate','netIncome','otherOperatingExpenses','costOfRevenue','grossProfit','totalRevenue']]
        df_balanco= df_balanco[['endDate','cash','shortTermInvestments','netReceivables','inventory','providers','longTermLoansAndFinancing','totalCurrentAssets','currentLiabilities','totalAssets']]
        df_estatistica= df_estatistica[['endDate','sharesOutstanding']]
        df_caixa = df_caixa[['endDate','operatingCashFlow']]
        df_financeiro = df_financeiro[['endDate','returnOnAssets','currentRatio','grossMargins']]

        df_merge = df_balanco.merge(df_dre, on='endDate').merge(df_estatistica, on='endDate').merge(df_caixa, on='endDate').merge(df_financeiro, on='endDate').copy()

        # df_merge['capital_investido'] = df_merge['totalCurrentAssets'].fillna(0) - df_merge['currentLiabilities'].fillna(0) + df_merge['propertyPlantEquipment'].fillna(0)
        # df_merge['ROC'] = df_merge['ebit'] / df_merge['capital_investido']
        # df_merge['EY']  = df_merge['ebit'] / df_merge['enterpriseValue'] # 
        
        dicio_merge = {
            'ativo':filename,
            'data':df_merge
        }
        lista_df.append(dicio_merge)
    except Exception as e:
        print(e)
        continue
```

```python
from matplotlib import axis
from scipy.datasets import ascent


for i in range(len(lista_df)):
    # print(lista_df[i]['ativo'])
    ativo = lista_df[i]['ativo']
    df = lista_df[i]['data']
    # df = df.sort_index(axis=0, ascending=False)
    df = df[::-1]
    #   grupo 1 - Rentabilidade - Lucratividade
    df['ROA'] = (df['returnOnAssets']>0).astype(int)
    df['Qualidade_ROA'] = (df['returnOnAssets'].diff(1)>0).astype(int)
    df['FCO'] = (df['operatingCashFlow']>0).astype(int)
    df['Qualidade_FCO'] = (df['operatingCashFlow'].diff(1)>0).astype(int)

    #   grupo 2 - Alavancagem, Liquidez e Financiamento
    df['Reducao_DIVIDA'] = (df['longTermLoansAndFinancing'].diff(1)>0).astype(int)
    df['Liquidez_CORRENTE'] = (df['currentRatio'].diff(1)>0).astype(int)
    df['Diluicao'] = (df['sharesOutstanding'].diff(1)>0).astype(int)

    #   grupo 3 - Eficiência Operacional (Margem bruta, Giro de Ativos)
    df['Margem_BRUTA'] = (df['grossMargins'].diff(1)>0).astype(int)
    df['Giro_ATIVOS_conta'] = df['totalRevenue'] / df['totalAssets'] 
    df['Giro_ATIVOS'] = (df['Giro_ATIVOS_conta'].diff(1)>0).astype(int)

    #   TOTAL 9 
    
    df[ativo] = df[['ROA','Qualidade_ROA','FCO','Qualidade_FCO','Reducao_DIVIDA','Liquidez_CORRENTE','Diluicao','Margem_BRUTA','Giro_ATIVOS']].sum(axis=1)

    df = df[['endDate',ativo]]
    df=df[::-1]

    if len(df)>=10:
        df.to_csv(f'piotroski/{ativo}')
```
